```python
#calculate a and b for butterworht filter
import numpy as np
import scipy.signal as signal

def butterworth_coefficients(order, cutoff_freq, sample_rate):
    """
    Calculate the coefficients for a Butterworth filter.

    Parameters:
    order (int): The order of the filter.
    cutoff_freq (float): The cutoff frequency in Hz.
    sample_rate (float): The sample rate in Hz.

    Returns:
    tuple: A tuple containing the numerator (b) and denominator (a) coefficients of the filter.
    """
    nyquist = 0.5 * sample_rate
    normalized_cutoff = cutoff_freq / nyquist
    b, a = signal.butter(order, normalized_cutoff, btype='low', analog=False)
    return b, a

if __name__ == "__main__":
    b, a = butterworth_coefficients(4, 40, 200)
    print(b, a)
```
